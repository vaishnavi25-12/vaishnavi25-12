```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("hojjatk/mnist-dataset")

print("Path to dataset files:", path)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
```

```python
mnist = tf.keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize the data
x_train = x_train / 255.0
x_test = x_test / 255.0
```

```python
model = Sequential()

# Input layer
model.add(Flatten(input_shape=(28,28)))

# Hidden layer 1
model.add(Dense(128, activation='relu'))

# Hidden layer 2 (added layer)
model.add(Dense(64, activation='relu'))

# Output layer
model.add(Dense(10, activation='softmax'))
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(x_train, y_train, epochs=5)
```

```python
model.evaluate(x_test, y_test)
```

# Neural Network for Handwritten Digit Classification using MNIST

 1. Introduction

This project demonstrates how to build a simple neural network model to classify handwritten digits using the MNIST dataset. The objective is to understand how neural networks work and how adding hidden layers can improve the model's ability to learn patterns from data.

Neural networks are widely used in machine learning and deep learning applications such as image recognition, speech recognition, and natural language processing.


2. Objective

The main objectives of this experiment are:

* To understand the structure of a neural network.
* To load and preprocess the MNIST dataset.
* To build a neural network model using TensorFlow/Keras.
* To add hidden layers to the neural network.
* To train the model and evaluate its accuracy.



3. Dataset Description

The MNIST dataset is a widely used dataset for image classification tasks.

Dataset details:

* Total images: 70,000 handwritten digits
* Training images: 60,000
* Testing images: 10,000
* Image size: 28 × 28 pixels
* Number of classes: 10 (digits from 0 to 9)

Each image represents a handwritten digit, and the model learns to classify which digit it represents.

---

4. Data Preprocessing

Before training the neural network, the dataset is preprocessed.

Steps involved:

1. Load the dataset using TensorFlow.
2. Normalize pixel values by dividing by 255.
3. Split the dataset into training and testing sets.

Normalization helps improve the training efficiency of the neural network.



 5. Neural Network Architecture

The neural network used in this project contains multiple layers.

Architecture:

Input Layer
↓
Flatten Layer (Converts 2D image to 1D vector)
↓
Dense Layer (128 neurons, ReLU activation)
↓
Dense Layer (64 neurons, ReLU activation)
↓
Dense Layer (32 neurons, ReLU activation)
↓
Output Layer (10 neurons, Softmax activation)

Explanation:

* **Flatten Layer** converts the 28×28 image into a vector of 784 values.
* **Dense Layers** act as hidden layers that learn patterns from the dataset.
* **ReLU activation** helps the model learn complex relationships.
* **Softmax activation** is used in the output layer for multi-class classification.



 6. Training the Model

The model is trained using the following settings:

* Optimizer: Adam
* Loss Function: Sparse Categorical Crossentropy
* Metrics: Accuracy
* Epochs: 5

During training, the neural network adjusts its weights to minimize the loss function and improve prediction accuracy.



7. Model Evaluation

After training, the model is evaluated on the test dataset.

Evaluation metrics include:

* Test Loss
* Test Accuracy

The accuracy shows how well the model can correctly classify unseen handwritten digits.



 8. Effect of Adding Layers

Adding hidden layers increases the learning capacity of the neural network. Deeper networks can capture more complex patterns in data.

However, too many layers may increase training time and lead to overfitting if not handled properly.



9. Result

The neural network successfully learns to classify handwritten digits from the MNIST dataset with high accuracy.

The trained model achieves good performance on both training and testing datasets.



10. Conclusion

This project demonstrates the implementation of a neural network using the MNIST dataset. By adding hidden layers, the model becomes capable of learning complex patterns from image data.
